## Modelo 1

#### Treinar o modelo

```python
import re
import spacy
from gensim.models import Word2Vec

nlp = spacy.load("pt_core_news_sm")


ficheiros = [
    "Harry Potter e A Pedra Filosofal.txt", 
    "Harry_Potter_Camara_Secreta-br.txt"
]

sentences = []

for nome_ficheiro in ficheiros:
    print(f"A processar: {nome_ficheiro}...")
    
    with open(nome_ficheiro, "r", encoding="utf8") as f:
        texto = f.read()
        
        texto = re.sub(r'\f', '', texto)
        
        doc = nlp(texto)
        
        for sent in doc.sents:
            tokens = [token.text.lower() for token in sent if not token.is_punct and not token.is_space]
            if len(tokens) > 0:
                sentences.append(tokens)


print(f"Total de frases carregadas: {len(sentences)}")
model_1 = Word2Vec(sentences, vector_size=150, window=5, min_count=2, sg=0, epochs=20, workers=3)
```

```python
model_1.wv['harry']
```

```python
model_1.wv['laura']
```

```python
model_1.wv.save_word2vec_format('model_harry.txt',binary=False)
#projector.tensorflow.org
```

```python
!python -m gensim.scripts.word2vec2tensor -i model_harry.txt -o model_harry
```

```python
len(model_1.wv)
```

```python
model_1.wv.index_to_key
```

```python
vetor_varinha = model_1.wv['varinha']
```

```python
'harry' in model_1.wv
```

### Palavras semelhantes

```python
model_1.wv.most_similar('harry')
```

```python
model_1.wv.most_similar('voldemort')
```

```python
print(model_1.wv.similarity('harry', 'snape'))
print(model_1.wv.similarity('harry', 'hagrid'))
print(model_1.wv.similarity('rony', 'hermione'))
print(model_1.wv.similarity('harry', 'draco'))
```

```python
# Pares de palavras do universo Harry Potter
pairs = [
    ('harry', 'rony'),          # Amigos (deve ter valor alto)
    ('harry', 'hermione'),     # Amigos (deve ter valor alto)
    ('harry', 'voldemort'),    # Inimigos (contextos diferentes ou semelhantes?)
    ('dumbledore', 'snape'),   # Professores
    ('varinha', 'vassoura'),   # Objetos mágicos
    ('grifinória', 'sonserina') # Casas (usa 'grifinória' e 'sonserina' se o livro for PT-BR)
]

for w1, w2 in pairs:
    print('%s \t %s \t% .2f' % (w1, w2, model_1.wv.similarity(w1,w2)))
```

### Intruso

```python
model_1.wv.doesnt_match(['harry', 'rony', 'hermione', 'draco']) #malfoy/draco
```

```python
#model.wv.doesnt_match(['rony', 'fred', 'george', 'harry'])
model_1.wv.doesnt_match(['weasley', 'fred', 'rony', 'potter'])
```

```python
model_1.wv.doesnt_match(['grifinória', 'sonserina', 'corvinal', 'lufa-lufa', 'vassoura'])
```

```python
resultado = model_1.wv.most_similar(positive=['sonserina', 'harry'], negative=['grifinória'])
print("Harry - Grifinória + Sonserina =")
print(resultado)
```

```python
resultado = model_1.wv.most_similar(positive=['harry', 'malfoy'], negative=['potter'])
print("Harry - Potter + Malfoy")
print(resultado)
```

```python
model_1.wv.most_similar("harry", topn=50)
```

### Analogia

```python
def analogy(x2,y1,x1):
    result = model_1.wv.most_similar(positive=[y1, x2], negative=[x1])
    return result[0][0]
```

```python
analogy('rony', 'draco', 'harry')
```

```python
analogy('sonserina', 'lufa-lufa', 'grifinória')
```

```python
analogy('escola', 'feitiço', 'varinha')
```

```python
analogy('lorde', 'snape', 'dumbledore')
```
